# Exception Handling in Python

## Definition

Exception handling is a mechanism in Python used to **handle runtime errors gracefully** without stopping the entire program.

It allows us to detect errors and execute alternative code instead of terminating the program.

---

## Basic Syntax

```python
try:
    # Code that may cause an error
except:
    # Code to handle the error

| Exception           | Meaning                       |
| ------------------- | ----------------------------- |
| `ValueError`        | Invalid value                 |
| `TypeError`         | Wrong data type               |
| `ZeroDivisionError` | Division by zero              |
| `IndexError`        | Index does not exist          |
| `KeyError`          | Dictionary key does not exist |
| `NameError`         | Variable does not exist       |
| `FileNotFoundError` | File does not exist           |
| `AttributeError`    | Invalid attribute/method      |
| `ImportError`       | Import operation failed       |


### Possible Combinations  

- `try` + `except`
- `try` + `except `+ `else`
- `try` + `except `+ `finally`
- `try` + `except `+ `else` + `finally`
- `try` + `finally`

## Key Points to Remember

- `try` → contains potentially risky code
- `except` → handles the exception
- `else` → runs when no exception occurs
- `finally` → always runs
- `raise` → manually raises an exception
- Prefer **specific exceptions** over a bare `except`
- Don't use exception handling to hide programming mistakes
- Keep the `try` block as small as practical

# Exception Handling – Quick Rules

- `try` → Code that may cause an exception.
- `except` → Runs **only when an exception occurs** and matches the exception type.
- `else` → Runs **only when no exception occurs** in `try`.
- `finally` → Runs **regardless of exception**; mainly used for cleanup.
- `raise` → Used to **manually raise an exception**.

## Execution Flow

```text
try
 ↓
Exception?
 ├── Yes → except → finally
 └── No  → else → finally

| Situation                 | `try` | `except` | `else` | `finally` |
| ------------------------- | ----- | -------- | ------ | --------- |
| No exception              | ✅     | ❌        | ✅      | ✅         |
| `ValueError`              | ⚠️    | ✅        | ❌      | ✅         |
| `ZeroDivisionError`       | ⚠️    | ✅        | ❌      | ✅         |
| Other unhandled exception | ⚠️    | ❌        | ❌      | ✅         |


````python
try:
    x = 10 / 0
except ZeroDivisionError:
    print("Error occurred")
````

````python
try:
    my_list = [1,2,3,4,5,6]

    print(f'{my_list[0]}')
    print(f'{my_list[1]}')
    print(f'{my_list[22]}')

    print(my_list[0]/0)

except IndexError:
    print('Code has error')
````

````python
try:
    my_list = [1,2,3,4,5,6]

    print(f'{my_list[0]}')
    print(f'{my_list[1]}')
    print(f'{my_list[2]}')

    print(my_list[0]/0)

except IndexError:
    print('Index out of range')

except ZeroDivisionError:
    print('num can not be divide by zero')
````
