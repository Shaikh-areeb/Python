# 🐍 Python Set

## 📌 Definition
A **Set** in Python is an **unordered and mutable collection** used to store **unique values**.

- Duplicate values are **not allowed**
- Elements are stored in **no specific order**
- Supports **different data types**
- Provides **very fast lookup operations** using hashing

Example:
```python
my_set = {10, 20, 30, "Python", 5.5}
```

---

# ⚙️ Important Set Methods

| Method | Purpose |
|----------|----------|
| `add(x)` | Adds a single element |
| `update()` | Adds multiple elements |
| `remove(x)` | Removes an element (raises error if not found) |
| `discard(x)` | Removes an element (no error if not found) |
| `pop()` | Removes a random element |
| `clear()` | Removes all elements |
| `copy()` | Creates a copy of the set |
| `union()` | Combines two sets |
| `intersection()` | Returns common elements |
| `difference()` | Returns elements present only in first set |
| `symmetric_difference()` | Returns elements present in either set but not both |

---

# 📏 Basic Rules

- Sets are **mutable**
- Duplicate values are **automatically removed**
- **Indexing and slicing are not supported**
- Elements are **unordered**
- Only **hashable (immutable)** objects can be stored
- Nested lists, dictionaries, and sets are not allowed as elements
- Membership testing (`in`) is very fast

---

# 🧠 Useful Tips & Tricks

### 1️⃣ Create an Empty Set
```python
my_set = set()
```

⚠️ `{}` creates a dictionary, not a set.

### 2️⃣ Remove Duplicates from a List
```python
unique_values = list(set(my_list))
```

### 3️⃣ Check Membership Efficiently
```python
10 in my_set
```

### 4️⃣ Find Common Elements
```python
set1.intersection(set2)
```

### 5️⃣ Find Unique Elements Between Sets
```python
set1.difference(set2)
```

### 6️⃣ Merge Multiple Sets
```python
set1.union(set2)
```

### 7️⃣ Use Sets for Fast Lookup
Sets provide **O(1)** average time complexity for searching elements.

---

# 🚀 Time Complexity

| Operation | Average Time Complexity |
|------------|------------------------|
| Add Element | O(1) |
| Remove Element | O(1) |
| Search Element | O(1) |
| Union | O(n + m) |
| Intersection | O(min(n, m)) |
| Difference | O(n) |

---

# 🎯 Interview Points

- Sets are implemented using **Hash Tables**.
- Duplicate values are automatically removed.
- Sets are unordered and do not support indexing.
- Membership testing is faster in sets than lists.
- `remove()` raises an error if the element is absent, whereas `discard()` does not.
- Empty set is created using `set()`, not `{}`.
