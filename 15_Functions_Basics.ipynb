```python
# Types Of Functions
"""
1. without parameters , without return
2. with parameters , without return
3. without parameters , with return
4. with parameters , with return

"""
```

```python
# Scoping

def addition():
    # Scoping
    num1 = int(input("enter the num 1 = "))
    num2 = int(input("enter the num 2 = "))
    # num1 and num2 are only valid in inside this func 
    # num1 and num2 cannot be accessed outside the func
    print(f'Answer = {num1 + num2}')
addition()
```

```python
# 1. without parameters , without return

def greet():            # without parameters
    print("hello")      # without return
greet()
```

```python
# 2. with parameters , without return

def add(n1, n2):        # with parameters 
    total = n1 + n2     # without return
    print(total)
add(10,20)              # Arguments 

# no of arguments == no of parameters
```

```python
# 3. without parameters , with return

def sub():                                          # without parameters
    num1 = int(input("enter the num 1 = "))            
    num2 = int(input("enter the num 2 = "))
    return num1 - num2                              # with return
sub()
```

```python
def division(num1, num2):                           # with parameters
    total = num1/num2
    return total                                    # with return
division(100,10)
```

```python
def reverse(str):
    rev = str[::-1]
    return rev

name = "Areeb"
reverse(name)
```

```python
def counter(str):
    d = {"upper":0 , "lower":0}
    for i in str:
        if i.isupper():
            d["upper"] += 1
        else:
            d["lower"] += 1
    return  d
    

name = "Areeb Shaikh"
counter(name)
```

```python
def distinct(a):
    return set(a)

lst = [1,1,2,2,4,5,5,6,7,65,4,3,2]
distinct(lst)
```

```python
def prime(num):
    factors = 0
    for i in range(1,num+1):
        if num % i == 0:
            factors += 1
    if factors == 2 :
        print("Prime")
    else:
        print("not prime")

num = int(input("enter the number"))
print(f'your number is {num}')
prime(num)

```

```python
def primeCount(start,end):
    prime = []
    for i in range(start,end +1):
        if i > 1:
            for j in range(2, int(i**0.5) + 1):
                if i % j == 0:
                    break
            else:
                prime.append(i)
            
    print(f'prime nos are = {prime}')

s = 2 
e = 50
primeCount(s,e)
```

```python
def Evenlen(word):
    words = word.split()
    for i in words:
        if len(i) % 2 == 0:
            print(word)

a = "hello how are you Areeb shaikh"
Evenlen(a)
```

```python
def summ(lst):
    total = 0
    for i in lst:
        total += i
    return total

lst = [1,2,3,4,5]
summ(lst)
```
